```python
# ============================================================
# CELL 9 — DISCOVER ALL BOOK / CHAPTER PAGES
# ============================================================

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, unquote
import re
import time

print("=" * 80)
print("DISCOVERING BOOK PAGES")
print("=" * 80)

# ------------------------------------------------------------
# Request headers
# ------------------------------------------------------------

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0 Safari/537.36"
    )
}

# ------------------------------------------------------------
# Download main book page
# ------------------------------------------------------------

print("\nOpening book page...")

response = requests.get(
    BOOK_URL,
    headers=headers,
    timeout=30
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise Exception(
        f"Could not access Wikisource page. "
        f"HTTP status: {response.status_code}"
    )

print("✓ Main page downloaded")

# ------------------------------------------------------------
# Parse HTML
# ------------------------------------------------------------

soup = BeautifulSoup(response.text, "html.parser")

print("Page title:", soup.title.get_text(strip=True) if soup.title else "No title")

# ------------------------------------------------------------
# Find all links
# ------------------------------------------------------------

all_links = soup.find_all("a", href=True)

print("Total links found on page:", len(all_links))

# ------------------------------------------------------------
# Extract Wikisource book links
# ------------------------------------------------------------

discovered_pages = []

for link in all_links:

    href = link.get("href", "").strip()
    text = link.get_text(" ", strip=True)

    if not href:
        continue

    # Convert relative URL to absolute URL
    full_url = urljoin(BASE_WIKISOURCE, href)

    # Only Wikisource wiki pages
    if not full_url.startswith(BASE_WIKISOURCE + "/wiki/"):
        continue

    # Remove fragments
    full_url = full_url.split("#")[0]

    # Ignore special/non-content namespaces
    wiki_part = full_url.split("/wiki/", 1)[1]

    if ":" in wiki_part:
        continue

    # Ignore empty links
    if not text and not wiki_part:
        continue

    # Avoid duplicates
    if full_url not in [x["url"] for x in discovered_pages]:

        discovered_pages.append({
            "title": text if text else unquote(wiki_part),
            "url": full_url
        })

# ------------------------------------------------------------
# If no links found, show diagnostic information
# ------------------------------------------------------------

if len(discovered_pages) == 0:

    print("\n⚠ No book pages were discovered.")

    print("\nFirst 20 wiki links found on the page:")

    count = 0

    for link in all_links:

        href = link.get("href", "")

        if "/wiki/" in href:
            print(
                f"{count + 1}. "
                f"{link.get_text(' ', strip=True)[:80]} "
                f"-> {href}"
            )

            count += 1

            if count >= 20:
                break

    print("\nPossible causes:")
    print("1. BOOK_URL is incorrect.")
    print("2. The book page uses a different Wikisource structure.")
    print("3. The chapter links are generated differently.")

else:

    print("\n" + "=" * 80)
    print("DISCOVERED BOOK PAGES")
    print("=" * 80)

    print("Total:", len(discovered_pages))

    for i, page in enumerate(discovered_pages[:30], 1):

        print(f"{i}. {page['title']}")
        print(f"   {page['url']}")

    if len(discovered_pages) > 30:
        print(
            f"\n... and {len(discovered_pages) - 30} more pages"
        )

print("\n" + "=" * 80)
```

```python
# ============================================================
# CELL 10 — ROBUST WIKISOURCE PAGE DISCOVERY
# ============================================================

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, parse_qs, unquote
import re

print("=" * 80)
print("ROBUST WIKISOURCE PAGE DISCOVERY")
print("=" * 80)

# ------------------------------------------------------------
# Request headers
# ------------------------------------------------------------

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0 Safari/537.36"
    ),
    "Accept-Language": "bn,en;q=0.9"
}

# ------------------------------------------------------------
# Download main book page
# ------------------------------------------------------------

print("\n1. Downloading main book page...")

response = requests.get(
    BOOK_URL,
    headers=headers,
    timeout=30
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise Exception(
        f"Failed to access Wikisource page: "
        f"HTTP {response.status_code}"
    )

print("✓ Page downloaded successfully")

# ------------------------------------------------------------
# Parse HTML
# ------------------------------------------------------------

soup = BeautifulSoup(response.text, "html.parser")

page_title = soup.title.get_text(" ", strip=True) if soup.title else ""

print("Page title:", page_title)

# ------------------------------------------------------------
# Find main content area
# ------------------------------------------------------------

content = soup.select_one("#mw-content-text")

if content is None:
    content = soup

print("✓ Main content area located")

# ------------------------------------------------------------
# Discover all links
# ------------------------------------------------------------

links = content.find_all("a", href=True)

print("Links inside main content:", len(links))

# ------------------------------------------------------------
# Helper function
# ------------------------------------------------------------

def normalize_wikisource_url(href):

    if not href:
        return None

    href = href.strip()

    # Ignore unwanted links
    if href.startswith("#"):
        return None

    if href.startswith("javascript:"):
        return None

    if href.startswith("mailto:"):
        return None

    # Convert relative URL
    url = urljoin(BASE_WIKISOURCE, href)

    # Remove fragment
    url = url.split("#")[0]

    parsed = urlparse(url)

    # Only Bengali Wikisource
    if parsed.netloc != urlparse(BASE_WIKISOURCE).netloc:
        return None

    # --------------------------------------------------------
    # Standard /wiki/ URL
    # --------------------------------------------------------

    if parsed.path.startswith("/wiki/"):

        title = unquote(
            parsed.path[len("/wiki/"):]
        )

        return url, title

    # --------------------------------------------------------
    # MediaWiki index.php URL
    # --------------------------------------------------------

    if parsed.path == "/w/index.php":

        query = parse_qs(parsed.query)

        if "title" in query:

            title = query["title"][0]

            clean_url = (
                BASE_WIKISOURCE
                + "/wiki/"
                + title.replace(" ", "_")
            )

            return clean_url, title

    return None


# ------------------------------------------------------------
# Discover candidate pages
# ------------------------------------------------------------

candidate_pages = []

for link in links:

    href = link.get("href", "")
    text = link.get_text(" ", strip=True)

    result = normalize_wikisource_url(href)

    if result is None:
        continue

    url, title = result

    # Ignore special namespaces
    if ":" in title:
        continue

    # Ignore the main book page itself
    if url.rstrip("/") == BOOK_URL.rstrip("/"):
        continue

    # Ignore obvious non-book pages
    lower_title = title.lower()

    ignored_words = [
        "বিশেষ:",
        "সাহায্য:",
        "উইকিসোর্স:",
        "শ্রেণী:",
        "টেমপ্লেট:",
        "পাতা:",
        "ব্যবহারকারী:"
    ]

    if any(word.lower() in lower_title for word in ignored_words):
        continue

    # Save unique pages
    if not any(p["url"] == url for p in candidate_pages):

        candidate_pages.append({
            "title": text if text else title,
            "url": url,
            "wiki_title": title
        })


# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("DISCOVERED CANDIDATE BOOK PAGES")
print("=" * 80)

print("Total:", len(candidate_pages))

for i, page in enumerate(candidate_pages, 1):

    print(f"\n{i}. {page['title']}")
    print("   Wiki title:", page["wiki_title"])
    print("   URL:", page["url"])

    if i >= 50:
        print(
            f"\n... showing first 50 of "
            f"{len(candidate_pages)} pages"
        )
        break


# ------------------------------------------------------------
# Diagnostic: show Bengali-looking links
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("BENGALI LINK DIAGNOSTIC")
print("=" * 80)

bengali_links = []

for link in links:

    href = link.get("href", "")
    text = link.get_text(" ", strip=True)

    # Detect Bengali Unicode characters
    if re.search(r"[\u0980-\u09FF]", text):

        bengali_links.append({
            "text": text,
            "href": href
        })

print("Bengali-text links found:", len(bengali_links))

for i, item in enumerate(bengali_links[:30], 1):

    print(f"{i}. {item['text'][:100]}")
    print(f"   {item['href']}")

print("\n" + "=" * 80)

if len(candidate_pages) == 0:

    print("⚠ No candidate subpages discovered.")

    print("\nThis means the book may use one of these structures:")
    print("• A single long page")
    print("• Wikisource Page namespace")
    print("• Transclusion/template structure")
    print("• Different subpage navigation")

else:

    print(
        f"✓ Successfully discovered "
        f"{len(candidate_pages)} candidate pages."
    )
```

```python
# ============================================================
# CELL 12 — DOWNLOAD ALL DISCOVERED WIKISOURCE PAGES
# ============================================================

import requests
import time

print("=" * 80)
print("DOWNLOADING BOOK PAGES")
print("=" * 80)

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0 Safari/537.36"
    ),
    "Accept-Language": "bn,en;q=0.9"
}

raw_pages = []

for i, page in enumerate(candidate_pages, 1):

    print(f"\n[{i}/{len(candidate_pages)}] {page['title']}")

    try:

        response = requests.get(
            page["url"],
            headers=headers,
            timeout=30
        )

        print("Status:", response.status_code)

        if response.status_code == 200:

            raw_pages.append({
                "title": page["title"],
                "url": page["url"],
                "html": response.text,
                "wiki_title": page["wiki_title"]
            })

            print("✓ Downloaded")

        else:
            print("⚠ Failed")

    except Exception as e:

        print("⚠ Error:", str(e))

    time.sleep(0.5)


print("\n" + "=" * 80)
print("DOWNLOAD SUMMARY")
print("=" * 80)

print("Candidate pages:", len(candidate_pages))
print("Successfully downloaded:", len(raw_pages))
print("Failed:", len(candidate_pages) - len(raw_pages))
```

```python
# ============================================================
# CELL 13 — EXTRACT TEXT FROM WIKISOURCE PAGES
# ============================================================

from bs4 import BeautifulSoup

print("=" * 80)
print("EXTRACTING BOOK TEXT")
print("=" * 80)

extracted_pages = []

for i, page in enumerate(raw_pages, 1):

    print(f"[{i}/{len(raw_pages)}] {page['title']}")

    soup = BeautifulSoup(page["html"], "html.parser")

    # Main Wikisource content
    content = soup.select_one("#mw-content-text")

    if content is None:
        print("  ⚠ Main content not found")
        continue

    # Remove unwanted HTML elements
    for tag in content.select(
        "script, style, table, sup, nav, footer, "
        ".mw-editsection, .noprint, .navbox, "
        ".metadata, .mw-cite-backlink"
    ):
        tag.decompose()

    text = content.get_text(
        "\n",
        strip=True
    )

    if not text.strip():
        print("  ⚠ Empty text")
        continue

    extracted_pages.append({
        "title": page["title"],
        "url": page["url"],
        "text": text,
        "wiki_title": page["wiki_title"]
    })

    print("  Characters:", len(text))

print("\n" + "=" * 80)
print("TEXT EXTRACTION SUMMARY")
print("=" * 80)

print("Pages extracted:", len(extracted_pages))

if extracted_pages:
    print(
        "Total characters:",
        sum(len(x["text"]) for x in extracted_pages)
    )
```

```python
# ============================================================
# CELL 14 — CLEAN AND PREPROCESS BOOK TEXT
# ============================================================

import re

print("=" * 80)
print("CLEANING BOOK TEXT")
print("=" * 80)


def clean_bengali_text(text):

    # Normalize different newline characters
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Remove excessive spaces
    text = re.sub(r"[ \t]+", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    # Remove leading/trailing whitespace
    text = text.strip()

    return text


cleaned_pages = []

for page in extracted_pages:

    cleaned_text = clean_bengali_text(page["text"])

    if len(cleaned_text) < 50:
        continue

    cleaned_pages.append({
        "title": page["title"],
        "url": page["url"],
        "text": cleaned_text,
        "wiki_title": page["wiki_title"]
    })


print("Original pages:", len(extracted_pages))
print("Cleaned pages:", len(cleaned_pages))

print(
    "Total cleaned characters:",
    sum(len(x["text"]) for x in cleaned_pages)
)

print("\n✓ Text preprocessing completed")
```

```python
# ============================================================
# CELL 15 — CREATE METADATA
# ============================================================

print("=" * 80)
print("CREATING BOOK METADATA")
print("=" * 80)

book_metadata_pages = []

for index, page in enumerate(cleaned_pages, 1):

    metadata = {
        "book": BOOK_NAME,
        "author": AUTHOR,
        "edition": EDITION,
        "part": f"Page {index}",
        "chapter": page["title"],
        "section": page["title"],
        "source_url": page["url"],
        "page_number": index
    }

    book_metadata_pages.append({
        "text": page["text"],
        "metadata": metadata
    })


print("Documents created:", len(book_metadata_pages))

if book_metadata_pages:

    print("\nExample metadata:")
    print(book_metadata_pages[0]["metadata"])
```

```python
# ============================================================
# CELL 16 — CREATE LANGCHAIN DOCUMENT OBJECTS
# ============================================================

from langchain_core.documents import Document

print("=" * 80)
print("CREATING LANGCHAIN DOCUMENTS")
print("=" * 80)

documents = []

for item in book_metadata_pages:

    doc = Document(
        page_content=item["text"],
        metadata=item["metadata"]
    )

    documents.append(doc)


print("Total LangChain Documents:", len(documents))

if documents:

    print("\nFirst document:")
    print("-" * 80)

    print(documents[0].page_content[:1000])

    print("\nMetadata:")
    print(documents[0].metadata)

print("\n✓ LangChain documents ready")
```

```python
# ============================================================
# CELL 17 — CHUNKING STRATEGY A
# ============================================================

from langchain_text_splitters import RecursiveCharacterTextSplitter

print("=" * 80)
print("CHUNKING STRATEGY A")
print("=" * 80)

splitter_A = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=150,
    separators=[
        "\n\n",
        "\n",
        "। ",
        "৷ ",
        "? ",
        "! ",
        " ",
        ""
    ]
)

chunks_A = splitter_A.split_documents(documents)

print("Original documents:", len(documents))
print("Chunks created:", len(chunks_A))

if chunks_A:

    print("\nExample chunk:")
    print("-" * 80)
    print(chunks_A[0].page_content[:1000])

    print("\nMetadata:")
    print(chunks_A[0].metadata)
```

```python
# ============================================================
# CELL 18 — CHUNKING STRATEGY B
# ============================================================

splitter_B = RecursiveCharacterTextSplitter(
    chunk_size=1200,
    chunk_overlap=200,
    separators=[
        "\n\n",
        "\n",
        "। ",
        "৷ ",
        "? ",
        "! ",
        " ",
        ""
    ]
)

chunks_B = splitter_B.split_documents(documents)

print("=" * 80)
print("CHUNKING STRATEGY B")
print("=" * 80)

print("Original documents:", len(documents))
print("Chunks created:", len(chunks_B))

if chunks_B:

    print("\nExample chunk:")
    print("-" * 80)

    print(chunks_B[0].page_content[:1200])

    print("\nMetadata:")
    print(chunks_B[0].metadata)
```

```python
# ============================================================
# CELL 19 — COMPARE CHUNKING STRATEGIES
# ============================================================

print("=" * 80)
print("CHUNKING STRATEGY COMPARISON")
print("=" * 80)

print("\nStrategy A")
print("Chunk size   : 800")
print("Overlap      : 150")
print("Total chunks :", len(chunks_A))

print("\nStrategy B")
print("Chunk size   : 1200")
print("Overlap      : 200")
print("Total chunks :", len(chunks_B))

print("\n✓ Both strategies created successfully")
```

```python
# ============================================================
# CELL 20 — BUILD FAISS VECTOR DATABASE
# ============================================================

from langchain_community.vectorstores import FAISS

print("=" * 80)
print("BUILDING FAISS VECTOR DATABASE")
print("=" * 80)

print("Using Strategy A")
print("Number of chunks:", len(chunks_A))

vectorstore = FAISS.from_documents(
    chunks_A,
    embeddings
)

print("\n✓ FAISS vector database created")

print(
    "Embedding dimension:",
    len(embeddings.embed_query("বাংলা সাহিত্য"))
)
```

```python
# ============================================================
# CELL 21 — SAVE FAISS VECTOR DATABASE
# ============================================================

import os

FAISS_DIR = "/content/book_faiss_db"

os.makedirs(
    FAISS_DIR,
    exist_ok=True
)

vectorstore.save_local(
    FAISS_DIR
)

print("=" * 80)
print("FAISS DATABASE SAVED")
print("=" * 80)

print("Location:", FAISS_DIR)

print("\nFiles:")

for file in os.listdir(FAISS_DIR):
    print("-", file)
```

```python
# ============================================================
# CELL 22 — LOAD FAISS DATABASE
# ============================================================

print("=" * 80)
print("LOADING FAISS DATABASE")
print("=" * 80)

vectorstore = FAISS.load_local(
    FAISS_DIR,
    embeddings,
    allow_dangerous_deserialization=True
)

print("✓ FAISS database loaded successfully")
```

```python
# ============================================================
# CELL 27 — BOOK-ONLY RAG PROMPT
# ============================================================

from langchain_core.prompts import ChatPromptTemplate

RAG_PROMPT = ChatPromptTemplate.from_template(
"""
আপনি একটি নির্দিষ্ট বাংলা বইভিত্তিক প্রশ্নোত্তর সহকারী।

আপনার একমাত্র জ্ঞানসূত্র হলো নিচে দেওয়া BOOK CONTEXT।

কঠোর নিয়ম:

1. শুধুমাত্র BOOK CONTEXT ব্যবহার করে উত্তর দিন।
2. নিজের সাধারণ জ্ঞান ব্যবহার করবেন না।
3. Context-এ উত্তর না থাকলে লিখুন:
   "এই তথ্যটি নির্বাচিত বইয়ে পাওয়া যায়নি।"
4. কোনো তথ্য অনুমান করবেন না।
5. কোনো চরিত্র, ঘটনা, স্থান বা তারিখ তৈরি করবেন না।
6. উত্তর বাংলা ভাষায় দিন।
7. উত্তর সংক্ষিপ্ত কিন্তু পরিষ্কার রাখুন।
8. উত্তরের শেষে source/chapter citation দিন।
9. Context-এর বাইরে কোনো তথ্য ব্যবহার করবেন না।

BOOK CONTEXT:
----------------
{context}
----------------

USER QUESTION:
{question}

ANSWER:
"""
)

print("✓ Strict book-only RAG prompt created")
```

```python
# ============================================================
# CELL 28 — FORMAT RETRIEVED DOCUMENTS
# ============================================================

def format_documents(docs):

    formatted = []

    for i, doc in enumerate(docs, 1):

        metadata = doc.metadata

        source_text = f"""
[Source {i}]
Book: {metadata.get('book', '')}
Author: {metadata.get('author', '')}
Edition: {metadata.get('edition', '')}
Chapter: {metadata.get('chapter', '')}
Section: {metadata.get('section', '')}
Source URL: {metadata.get('source_url', '')}

Text:
{doc.page_content}
"""

        formatted.append(source_text)

    return "\n\n".join(formatted)


print("✓ Document formatter ready")
```

```python
# ============================================================
# CELL 29 — MAIN RAG FUNCTION
# ============================================================

def ask_book(question, k=4):

    # --------------------------------------------------------
    # Step 1: Retrieve relevant book chunks
    # --------------------------------------------------------

    docs = retriever.invoke(question)

    # --------------------------------------------------------
    # Step 2: Handle no retrieval
    # --------------------------------------------------------

    if not docs:

        return {
            "answer": "এই তথ্যটি নির্বাচিত বইয়ে পাওয়া যায়নি।",
            "sources": []
        }

    # --------------------------------------------------------
    # Step 3: Format context
    # --------------------------------------------------------

    context = format_documents(docs)

    # --------------------------------------------------------
    # Step 4: Create prompt
    # --------------------------------------------------------

    messages = RAG_PROMPT.format_messages(
        context=context,
        question=question
    )

    # --------------------------------------------------------
    # Step 5: Ask LLM
    # --------------------------------------------------------

    response = llm.invoke(messages)

    answer = response.content

    # --------------------------------------------------------
    # Step 6: Collect sources
    # --------------------------------------------------------

    sources = []

    for doc in docs:

        source = {
            "chapter": doc.metadata.get("chapter", ""),
            "section": doc.metadata.get("section", ""),
            "url": doc.metadata.get("source_url", "")
        }

        if source not in sources:
            sources.append(source)

    return {
        "answer": answer,
        "sources": sources
    }


print("✓ Main RAG function created")
```

```python
# ============================================================
# CELL 32 — SCORED RETRIEVAL FUNCTION
# ============================================================

def retrieve_with_scores(
    query,
    k=4
):

    results = vectorstore.similarity_search_with_score(
        query,
        k=k
    )

    return results


print("✓ Scored retrieval function ready")
```

```python
# ============================================================
# CELL 33 — OUT-OF-BOOK QUERY CHECK
# ============================================================

def check_relevance(
    query,
    k=4,
    threshold=1.0
):

    results = retrieve_with_scores(
        query,
        k=k
    )

    if not results:
        return False, []

    best_score = results[0][1]

    # Lower FAISS distance generally means closer match
    is_relevant = best_score <= threshold

    return is_relevant, results


test_query = "বাংলাদেশের বর্তমান রাষ্ট্রপতির নাম কী?"

is_relevant, results = check_relevance(
    test_query
)

print("=" * 80)
print("OUT-OF-BOOK TEST")
print("=" * 80)

print("Question:", test_query)

if results:
    print("Best distance:", results[0][1])

print("Relevant:", is_relevant)

print("\nNote:")
print("The threshold must be calibrated using your actual book data.")
```

```python
# ============================================================
# CELL 34 — STRICT BOOK-ONLY RAG
# ============================================================

def strict_book_rag(
    question,
    k=4,
    threshold=1.0
):

    results = retrieve_with_scores(
        question,
        k=k
    )

    if not results:

        return {
            "answer": "এই তথ্যটি নির্বাচিত বইয়ে পাওয়া যায়নি।",
            "sources": []
        }

    best_score = results[0][1]

    # Reject weak retrieval
    if best_score > threshold:

        return {
            "answer": "এই তথ্যটি নির্বাচিত বইয়ে পাওয়া যায়নি।",
            "sources": []
        }

    docs = [doc for doc, score in results]

    context = format_documents(docs)

    messages = RAG_PROMPT.format_messages(
        context=context,
        question=question
    )

    response = llm.invoke(messages)

    sources = []

    for doc in docs:

        source = {
            "chapter": doc.metadata.get("chapter"),
            "section": doc.metadata.get("section"),
            "url": doc.metadata.get("source_url")
        }

        if source not in sources:
            sources.append(source)

    return {
        "answer": response.content,
        "sources": sources,
        "best_score": best_score
    }


print("✓ Strict book-only RAG created")
```
